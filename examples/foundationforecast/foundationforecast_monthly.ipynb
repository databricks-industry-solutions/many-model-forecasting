### Prepare data
We use the [`datasetsforecast`](https://github.com/Nixtla/datasetsforecast/tree/main/) package to download the M5 files and build the tables ourselves. `M5.load` itself fails on current runtimes, because it relies on a numba feature that newer numba releases removed. We keep the 500 series (one item in one store) with the highest sales over the last year, and start each series once the item has a price and has sold at least once.

We aggregate the daily rows to month ends: sales are summed, the price is averaged, and event days are counted. Months with missing days are dropped. SNAP days are not used here: every state has ten of them each month, so the monthly count never changes. The calendar ends in mid-June 2016, so the last complete months with known covariates are March to May 2016. They become the scoring table, with their sales withheld, and the months up to February 2016 are the training table.

All series end on the same date: MMF's data quality check removes series whose dates stop before the latest date in the dataset.

```python
n = 500  # Number of time series
freq = "M"
prediction_length = 3
covariates = ["sell_price", "event_days"]


def load_m5_daily(n):
    """Daily sales and covariates of the n best-selling M5 series (one item in one store).

    Sales are known up to 2016-05-22; the calendar and the prices continue to 2016-06-19,
    so the rows after the last sale carry known future covariates and a null ``y``.
    """
    directory = str(pathlib.Path.home())
    M5.download(directory)
    path = f"{directory}/m5/datasets"
    sales = pd.read_csv(f"{path}/sales_train_evaluation.csv")
    days = [c for c in sales.columns if c.startswith("d_")]
    # The best sellers over the last year have few zero days, which keeps sMAPE meaningful.
    sales = sales.loc[sales[days[-365:]].sum(axis=1).nlargest(n).index]
    sales["unique_id"] = sales["item_id"] + "_" + sales["store_id"]

    calendar = pd.read_csv(f"{path}/calendar.csv", parse_dates=["date"])
    # Row i of the calendar is day d_{i+1} of the sales file.
    calendar["d"] = [f"d_{i + 1}" for i in range(len(calendar))]
    calendar["event_days"] = calendar["event_name_1"].notna().astype(float)
    prices = pd.read_csv(f"{path}/sell_prices.csv").merge(
        sales[["item_id", "store_id"]], on=["item_id", "store_id"]
    )
    df = (
        sales[["unique_id", "item_id", "store_id", "state_id"]]
        .merge(calendar[["d", "date", "wm_yr_wk", "event_days", "snap_CA", "snap_TX", "snap_WI"]], how="cross")
        .merge(
            sales.melt(id_vars="unique_id", value_vars=days, var_name="d", value_name="y"),
            on=["unique_id", "d"], how="left",
        )
        .merge(prices, on=["store_id", "item_id", "wm_yr_wk"], how="left")
        .rename(columns={"date": "ds"})
        .sort_values(["unique_id", "ds"], ignore_index=True)
    )
    # SNAP (food stamp) days differ by state.
    df["snap_days"] = np.select(
        [df["state_id"] == "CA", df["state_id"] == "TX"], [df["snap_CA"], df["snap_TX"]], df["snap_WI"]
    ).astype(float)
    # Start each series once the item is listed (has a price) and has sold at least once.
    by_series = df.groupby("unique_id")
    listed = by_series["sell_price"].transform(lambda s: s.notna().cummax())
    sold = by_series["y"].transform(lambda s: (s > 0).cummax())
    df = df[listed & sold].copy()
    df["sell_price"] = df.groupby("unique_id")["sell_price"].ffill().astype(float)
    return df[["unique_id", "ds", "y", "sell_price", "snap_days", "event_days"]]


def to_frequency(df, freq):
    """Aggregate daily rows to weeks ending on Sunday (``W``) or to month ends (``M``).

    Sales are summed and prices averaged; SNAP and event days are counted. Periods with
    missing days are dropped, and a period's ``y`` is null unless every day's sales are known.
    """
    period = {"W": pd.offsets.Week(weekday=6), "M": pd.offsets.MonthEnd()}[freq]
    out = (
        df.groupby(["unique_id", pd.Grouper(key="ds", freq=period)])
        .agg(
            y=("y", "sum"), known_days=("y", "count"), days=("y", "size"),
            sell_price=("sell_price", "mean"), snap_days=("snap_days", "sum"), event_days=("event_days", "sum"),
        )
        .reset_index()
    )
    period_days = 7 if freq == "W" else out["ds"].dt.days_in_month
    out = out[out["days"] == period_days].copy()
    out.loc[out["known_days"] < out["days"], "y"] = np.nan
    return out[["unique_id", "ds", "y", "sell_price", "snap_days", "event_days"]]


def split_train_score(df, prediction_length):
    """The last ``prediction_length`` periods become the scoring rows: covariates only, no sales."""
    dates = np.sort(df["ds"].unique())
    cutoff = dates[-prediction_length - 1]
    train = df[df["ds"] <= cutoff].reset_index(drop=True)
    score = df[df["ds"] > cutoff].drop(columns="y").reset_index(drop=True)
    return train, score
```

```python
train, score = split_train_score(to_frequency(load_m5_daily(n), freq), prediction_length)
train = train[["unique_id", "ds", "y", *covariates]]
score = score[["unique_id", "ds", *covariates]]
example_id = train["unique_id"].min()

# Making sure that the catalog and the schema exist
if catalog not in [row[0] for row in spark.sql("SHOW CATALOGS").collect()]:
    _ = spark.sql(f"CREATE CATALOG {catalog}")
_ = spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{db}")

for table, pdf in [("ff_monthly_train", train), ("ff_monthly_score", score)]:
    (
        spark.createDataFrame(pdf)
        .write.format("delta").mode("overwrite").option("overwriteSchema", "true")
        .saveAsTable(f"{catalog}.{db}.{table}")
    )
```

The covariate models need a value for every covariate, every series and each of the next `prediction_length` dates after the series' last sale. Missing values, gaps and duplicate dates raise an error, so we check the tables before running anything:

```python
last_sale = train.groupby("unique_id")["ds"].max()
assert train[covariates].notna().all().all() and score[covariates].notna().all().all(), "covariates contain nulls"
assert set(score["unique_id"]) == set(last_sale.index), "some series have no scoring rows"
assert (score.groupby("unique_id").size() == prediction_length).all(), "each series needs prediction_length scoring rows"
assert (score.groupby("unique_id")["ds"].min() > last_sale).all(), "scoring rows must follow the last sale"
display(spark.sql(f"select count(distinct unique_id) as series, min(ds) as first_date, max(ds) as last_date from {catalog}.{db}.ff_monthly_train"))
display(spark.sql(f"select count(distinct unique_id) as series, min(ds) as first_date, max(ds) as last_date from {catalog}.{db}.ff_monthly_score"))
```

### Models
The FoundationForecast models start with `FF`. Before you run them, note:

### Run MMF

Now we run evaluation and forecasting with the `run_forecast` function. Refer to [README.md](https://github.com/databricks-industry-solutions/many-model-forecasting/blob/main/README.md#parameters-description) for a description of each parameter. The covariates are passed as `dynamic_future_numerical`, and `scoring_data` is the scoring table: its rows supply the covariate values for the forecast. During backtesting, the covariate values of each backtest window come from the training table. Each model is backtested, logged and registered to Unity Catalog under `model_output`, and then used to forecast. The same `run_id` is assigned to all models, which makes the post-evaluation analysis easier. On the first run, each checkpoint is downloaded from the Hugging Face Hub.

```python
from mmf_sa import run_forecast

run_id = str(uuid.uuid4())
settings = dict(
    spark=spark,
    train_data=f"{catalog}.{db}.ff_monthly_train",
    scoring_output=f"{catalog}.{db}.ff_monthly_scoring_output",
    evaluation_output=f"{catalog}.{db}.ff_monthly_evaluation_output",
    model_output=f"{catalog}.{db}",
    group_id="unique_id",
    date_col="ds",
    target="y",
    freq=freq,
    prediction_length=prediction_length,
    backtest_length=12,
    stride=1,
    metric="smape",
    train_predict_ratio=1,
    data_quality_check=True,
    resample=False,
    experiment_path=f"/Users/{user}/mmf/foundationforecast_monthly",
    run_id=run_id,
    accelerator="gpu",
    serverless=serverless,
)

for model in active_models:
    print(f"\n===== Running {model} with covariates (run_id={run_id}) =====")
    run_forecast(
        scoring_data=f"{catalog}.{db}.ff_monthly_score",
        dynamic_future_numerical=covariates,
        active_models=[model],
        use_case_name="ff_monthly",
        **settings,
    )
```

To see what the covariates add, we run the ten covariate models once more without them. These runs use `use_case_name="ff_monthly_no_covariates"`, so their results sit in the same output tables under a different `use_case`, and their models are registered under different names. Set `compare_without_covariates = False` to skip them.

```python
compare_without_covariates = True
covariate_models = [
    "FFChronos2",
    "FFChronos2Small",
    "FFTimesFM_3_0",
    "FFTiRex2",
    "FFT0Beta",
    "FFT0Gamma",
    "FFTimestral_5M",
    "FFTimestral_28M",
    "FFTimestral_123M",
    "FFTimestral_209M",
]

if compare_without_covariates:
    for model in covariate_models:
        print(f"\n===== Running {model} without covariates (run_id={run_id}) =====")
        run_forecast(
            scoring_data=f"{catalog}.{db}.ff_monthly_train",
            active_models=[model],
            use_case_name="ff_monthly_no_covariates",
            **settings,
        )
```

### Use a registered model
Each model is registered to Unity Catalog. A registered covariate model takes one table with the history rows followed by the future rows of each series, all with the covariate columns; the future rows have a null `y`. The example below loads the latest version of `FFChronos2` and forecasts three series.

```python
import mlflow
from mlflow import MlflowClient

mlflow.set_registry_uri("databricks-uc")
model_name = f"{catalog}.{db}.FFChronos2_ff_monthly"
version = max(int(v.version) for v in MlflowClient().search_model_versions(f"name = '{model_name}'"))
registered_model = mlflow.pyfunc.load_model(f"models:/{model_name}/{version}")

ids = sorted(train["unique_id"].unique())[:3]
model_input = pd.concat([
    train[train["unique_id"].isin(ids)],
    score[score["unique_id"].isin(ids)].assign(y=np.nan),
], ignore_index=True)
display(registered_model.predict(model_input))
```
